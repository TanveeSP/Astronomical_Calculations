```python
# Write your solution here
import numpy as np
import matplotlib.pyplot as plt

# Function 1: f(x) = e^x, x in [0, 5]
x1 = np.arange(0, 5.1, 0.1)
y1 = np.exp(x1)

# Function 2: f(x) = e^(-x), x in [0, 5]
x2 = np.arange(0, 5.1, 0.1)
y2 = np.exp(-x2)

# Function 3: f(x) = x * e^(-x), x in [0, 5]
x3 = np.arange(0, 5.1, 0.1)
y3 = x3 * np.exp(-x3)

# Function 4: f(x) = (1 / sqrt(2 * pi * sigma^2)) * exp(-x^2 / (2 * sigma^2)), x in [-5, 5]
sigmas = [0.5, 1.0, 2.0]  # Different values of sigma
x4 = np.arange(-5, 5.1, 0.1)

# Function 5: f(x) = sin^2(x), x in [0, 2*pi]
x5 = np.arange(0, 2 * np.pi + 0.1, 0.1)
y5 = np.sin(x5) ** 2

# Plotting
plt.figure(figsize=(15, 10))

# Plot Function 1
plt.subplot(3, 2, 1)
plt.plot(x1, y1, label='$e^x$')
plt.title('Function 1: $e^x$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()

# Plot Function 2
plt.subplot(3, 2, 2)
plt.plot(x2, y2, label='$e^{-x}$')
plt.title('Function 2: $e^{-x}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()

# Plot Function 3
plt.subplot(3, 2, 3)
plt.plot(x3, y3, label='$x e^{-x}$')
plt.title('Function 3: $x e^{-x}$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()

# Plot Function 4
plt.subplot(3, 2, 4)
for sigma in sigmas:
    y4 = (1 / np.sqrt(2 * np.pi * sigma**2)) * np.exp(-x4**2 / (2 * sigma**2))
    plt.plot(x4, y4, label=f'$\sigma={sigma}$')
plt.title('Function 4: Gaussian for different $\sigma$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()

# Plot Function 5
plt.subplot(3, 2, 5)
plt.plot(x5, y5, label='$\sin^2(x)$')
plt.title('Function 5: $\sin^2(x)$')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid()
plt.legend()

plt.tight_layout()
plt.show()



```
