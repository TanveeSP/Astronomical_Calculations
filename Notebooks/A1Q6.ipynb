```python
# Write your solution here
import numpy as np
import matplotlib.pyplot as plt

# Wien's displacement constant
b = 2.897e-3  # meters Kelvin

# Create a numpy array of temperatures
temperatures = np.array([1000, 2000, 3000, 4000, 5000, 6000, 7000])

# Calculate peak wavelengths directly using vectorized operations
peak_wavelengths = b / temperatures

# Print temperature and corresponding peak wavelengths
print("Temperature (K) | Peak Wavelength (m)")
for T, λ in zip(temperatures, peak_wavelengths):
    print(f"{T:<15} | {λ:.3e}")

# Create a smooth temperature range for plotting
smooth_temperatures = np.linspace(1000, 7000, 100)
smooth_wavelengths = b / smooth_temperatures

# Plot the data
plt.figure(figsize=(8, 5))
plt.plot(smooth_temperatures, smooth_wavelengths, label='Smooth Curve')
plt.scatter(temperatures, peak_wavelengths, color='red', label='Given Data Points')
plt.title('Temperature vs Peak Wavelength')
plt.xlabel('Temperature (K)')
plt.ylabel('Peak Wavelength (m)')
plt.legend()
plt.grid(True)
plt.show()


```
