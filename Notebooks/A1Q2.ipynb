```python
# Write your solution here
import numpy as np

def cartesian_to_spherical(x,y, z):

    # radial distance
    r = np.sqrt(x**2 + y**2 + z**2)

    # polar angle
    if r!=0:
       theta = np.arccos(z/r)
    else:
       theta = 0

    # azimuthal angle
    phi = np.arctan2(y,x)

    return r, theta, phi

# cartesian co-ordinates as input
x = float(input("Enter the x-coordinate: "))
y = float(input("Enter the y-coordinate: "))
z = float(input("Enter the z-coordinate: "))

#convert cartesian to sperical co-ordinates
r, theta, phi = cartesian_to_spherical(x,y,z)
print(f"Spherical Co-ordibates of Star: r={r}, theta={theta}radians, phi={phi}radians")




```
