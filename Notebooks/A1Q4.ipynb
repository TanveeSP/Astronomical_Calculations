```python
# Write your solution here
import numpy as np
import matplotlib.pyplot as plt

# Constants
g = 9.81  # Acceleration due to gravity (m/s^2)
v0 = 50   # Initial velocity (m/s)

# Function to calculate the range of a projectile
def calculate_range(v0, theta, g):
    theta_rad = np.radians(theta)  # Convert angle to radians
    return (v0**2 * np.sin(2 * theta_rad)) / g

# Create an array of launch angles from 0 to 80 degrees in steps of 10 degrees
angles = np.arange(0, 81, 10)

# Calculate the range for each angle
ranges = [calculate_range(v0, angle, g) for angle in angles]

# Plot angle vs range
plt.figure(figsize=(10, 6))
plt.plot(angles, ranges, marker='o', color='b', label="Range of Projectile")
plt.title("Projectile Range vs Launch Angle", fontsize=16)
plt.xlabel("Launch Angle (degrees)", fontsize=14)
plt.ylabel("Range (m)", fontsize=14)
plt.grid(True)
plt.legend(fontsize=12)
plt.show()


```
