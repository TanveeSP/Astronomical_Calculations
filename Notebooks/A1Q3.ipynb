```python
# Write your solution here
import numpy as np
def calculate_projectile_range(u, theta, g=9.8):
    theta_radians = np.radians(theta) # convert angle
    range_of_projectile = (u**2*np.sin(2*theta_radians))/g # calculate range
    return range_of_projectile

#user input
initial_velocity = float(input("Enter the initial velocity (u) in m/s: " ))
launch_angle = float (input("Enter the launch angle (theta) in degrees: "))
gravitational_acceleration = float(input("Enter the gravitatonal acceleration (g) in m/s**2 (default 9.8): "))

#calculate range of projectile
projectile_range = calculate_projectile_range(initial_velocity, launch_angle, gravitational_acceleration)

#output
print(f"For an initial velocity of {initial_velocity}m/s and "
      f"launch angle of {launch_angle} degrees, the range of the projectile is: "
      f"{projectile_range:.2f} meters.")


```
